### DATASETS

```python
# create dataset from nothing
ds = requests.post(url =  f"{baseurl}/datasets", 
                   json = dict(jsonable_encoder(Dataset())), 
                   headers = {"Authorization":f"Bearer {apikey}"})
try:
    print(ds.json())
except:
    print(ds.content)
```

```python
# create dataset from file upload
file_path = '/path/to/dir/'
file_name = 'wPMMA_FH_02.tif'
full_path = os.path.join(file_path, file_name)


# Send the file as bytes
file_payload = {'file': (file_name, open(full_path, 'rb')), 
                'filename':'file_name'}

ds = requests.post(url =  f"{baseurl}/datasets", 
                   json = dict(jsonable_encoder(Dataset())), 
                   files = file_payload,
                   headers = {"Authorization":f"Bearer {apikey}", 'Content-Type': 'multipart/form-data'})
try:
    print(ds.json())
except:
    print(ds.content)
```

```python
# create dataset from json
with open("test-data.json") as f:
    temp = json.load(f)

pyds = Dataset(temp)
```
