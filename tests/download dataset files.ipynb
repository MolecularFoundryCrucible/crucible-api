```python
def download_file(url,apikey, filename):
    try:
        response = requests.get(url, headers = {"Authorization":f"Bearer {apikey}"})
        response.raise_for_status()

        with open(filename, 'wb') as file:
            for chunk in response.iter_content(chunk_size=8192):
                file.write(chunk)
        print(f"File downloaded successfully to {filename}")
    
    except requests.exceptions.RequestException as e:
        print(f"An error occurred: {e}")
    except Exception as e:
         print(f"An unexpected error occurred: {e}")
        
# Example usage:
dsid = "0t0baae7bxttq000gyv16y9xrc"
filename = "YL_15_13_A3_0006.ibw"
filename = "0t0baae7bxttq000gyv16y9xrc.json"
url = f"{baseurl}/datasets/{dsid}/download/{filename}"
download_file(url, apikey, filename)
```
